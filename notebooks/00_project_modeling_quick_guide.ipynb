```python
from __future__ import annotations

import json
from pathlib import Path

import pandas as pd


def find_repo_root(start: Path) -> Path:
    start = start.resolve()
    for candidate in [start, *start.parents]:
        if (candidate / '.git').exists():
            return candidate
    raise FileNotFoundError('Could not find the repository root.')


REPO_ROOT = find_repo_root(Path.cwd())


def load_csv(rel_path: str) -> pd.DataFrame:
    return pd.read_csv(REPO_ROOT / rel_path)


def load_json(rel_path: str):
    return json.loads((REPO_ROOT / rel_path).read_text(encoding='utf-8'))


print('Repository root:', REPO_ROOT)
```

## 1. The project question

This repository uses **Black Marble nightlight data** to study a practical question:

> After a disaster, do pixels near critical infrastructure behave as if backup power or faster recovery is helping them stay brighter?

The active code separates that problem into three layers:

- **Explanatory mainline**: does the signal exist inside the observed sample?
- **Cross-event generalization**: does the signal transport to an unseen event?
- **Exploration and appendix lines**: what changes when we add hazard, quality, readiness, and BUG-specific logic?


```python
strict_summary = load_csv('project/modeling/output/model_summary_feature_upgrade_v2_strict.csv')
cross_summary = load_csv('project/modeling/output/model_summary_cross_event_v3.csv')
readiness = load_csv('project/modeling/output/event_readiness_score_v1.csv')

strict_focus = strict_summary.loc[
    strict_summary['variant'].eq('full_locked_v2_strict'),
    ['model', 'key_metric', 'value', 'p_value']
]

cross_focus = cross_summary.loc[
    cross_summary['model'].isin(['Logit', 'Cox', 'AFT']),
    ['model', 'metric_name', 'value', 'delta_vs_baseline']
]

print('Strict-v2 explanatory snapshot:')
print(strict_focus.to_string(index=False))
print()
print('Cross-event snapshot:')
print(cross_focus.to_string(index=False))
print()
print('Readiness snapshot:')
print(readiness.head(6).to_string(index=False))
```
